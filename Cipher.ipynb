## Rail fence cipher 

encrpyt

```python
def RF_encode(key,text=None):
    text_len = len(text)
    fences = []
    #create fences
    for fence in range(key):
        fences.append([])
    fence_count = len(fences)
    #add each consicutive element in a zig zag pattern in the sub-lists
    for i, char in enumerate(text):
        position = i % (2 * fence_count - 2)
        index = position if position < fence_count else 2 * fence_count - 2 - position
        fences[index].append(char)
    result = ''.join(''.join(sublist) for sublist in fences)

    return result
```

decrypt

```python
def RF_decode(string, n):
    lengths = [0] * n

    # Find how many characters belong to each sublist
    for i in range(len(string)):
        position = i % (2 * n - 2)
        index = position if position < n else 2 * n - 2 - position
        lengths[index] += 1

    # Split the jumbled string back into the sublists
    lists = []
    start = 0

    for length in lengths:
        lists.append(list(string[start:start + length]))
        start += length

    # Read the lists in zigzag order
    result = ""

    indexes = [0] * n

    for i in range(len(string)):
        position = i % (2 * n - 2)
        index = position if position < n else 2 * n - 2 - position

        result += lists[index][indexes[index]]
        indexes[index] += 1

    return result
RF_decode(RF_encode(4,"computer-science"), 4)
```
